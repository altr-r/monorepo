What is a dataframe in Pandas?
- A dataframe is a datastructure which is two dimensional and consisting of rows and columns. It is also size mutable and table like.
- The closest built in datastructure that we can use to understand dataframe is dcitionary of arrays. So a dictionary is a collection of key value pairs. Even though a dataframe is not actually a dictionary but we can use a dictionary to visiualize it.
- Think of it as an excel sheet

#### example 1

```python
car = {
  "brand": "Toyota",
  "model": "Premio",
  "year": 2019
}
```

#### example 2

```python
cars = {
  "brand": ["Toyota"],
  "model": ["Premio"],
  "year": [2019]
}
```

#### example 4

```python
cars = {
  "brand": ["Toyota", "Nissan"],
  "model": ["Premio", "Silvia"],
  "year": [2019, 2015]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(cars)

df
```
