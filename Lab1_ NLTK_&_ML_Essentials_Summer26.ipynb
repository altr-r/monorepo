```python
# Comparative analysis
from tabulate import tabulate

def stemming_example(words):
    porter = PorterStemmer()
    lancaster = LancasterStemmer()
    snowball = SnowballStemmer("english")

    data = [[word, porter.stem(word), lancaster.stem(word), snowball.stem(word)] for word in words]
    headers = ["Word", "PorterStemmer", "LancasterStemmer", "SnowballStemmer"]
    print("Stemming Results:")
    print(tabulate(data, headers=headers, tablefmt="grid"))
    print("\n")

def lemmatization_example(words):
    lemmatizer = WordNetLemmatizer()

    data = [[word, lemmatizer.lemmatize(word, pos='n'), lemmatizer.lemmatize(word, pos='v')] for word in words]
    headers = ["Word", "Lemmatized (noun)", "Lemmatized (verb)"]
    print("Lemmatization Results:")
    print(tabulate(data, headers=headers, tablefmt="grid"))
    print("\n")

words = ["connect", "connected", "connecting", "connection", "connections",
         "argue", "argued", "arguing", "argument", "arguments",
         "nation", "national", "nationality", "nationalism", "nationalist",
         "act", "acted", "acting", "action", "active", "activity",
         "derive", "derived", "deriving", "derivative", "derivation"]

print("Original Words:", words)
print("\n")

stemming_example(words)
lemmatization_example(words)
```

```python
import numpy as np

text = "Someone has told you that words in English can be separated by separated separated separated separated separated separated separated separated separated  simply splitting on whitespace. How many times would that heuristic fail for the following text?"
tokens = word_tokenize(text.lower())

unique_words = list(set(tokens))

#Initialize co-occurrence matrix (square matrix with zeros)
matrix = np.zeros((len(unique_words), len(unique_words)), dtype=int)

word_to_index = {}  #Create an empty dictionary

for i, word in enumerate(unique_words):
    word_to_index[word] = i  #Assigning index to the word

print(word_to_index)

for i in range(len(tokens)-1):
  current_word = tokens[i]
  next_word = tokens[i+1]
  cw_index = word_to_index[current_word]
  nw_index = word_to_index[next_word]
  matrix[cw_index,nw_index] += 1

print(matrix)
```

# Bag of Word

**Bag-of-Words (BoW)** is one of the earliest methods to convert words into vectors using the occurrences of the words in a document. It ignores the order or structure of the words.



bow.PNG

## Implementation from Scratch

Sample Corpus

```python
corpus = [
    "Cats chase mice in the garden garden.",
    "Dogs bark loudly at night.",
    "Birds sing sweet songs every morning.",
    "The cat sleeps on the warm sofa.",
    "A fast dog chased the red ball.",
    "Mice hide quickly when cats appear.",
]
```

```python

def create_bow_from_scratch(sentences):
    """
    Creates bag-of-words vectors and vocabulary from a list of sentences.

    Args:
        sentences (list): A list of strings, where each string is a sentence.

    Returns:
        tuple: A tuple containing:
            - bow_vectors (list of lists): A list where each inner list represents
              the BoW vector for a corresponding sentence.
            - vocabulary_dict (dict): A dictionary mapping unique words to their
              index in the vocabulary.
    """
    all_tokens = []
    for sentence in sentences:
        # 1. Convert corpuses to tokens using nltk word_tokenize()
        # Convert to lowercase for consistent word counting
        tokens = word_tokenize(sentence.lower())
        all_tokens.extend(tokens)

    # 2. Find unique words by converting it to set
    unique_words = set(all_tokens)

    # 3. Sort them in ascending order to create a consistent vocabulary
    vocabulary = sorted(list(unique_words))

    # Create vocabulary dictionary (word to index mapping)
    vocabulary_dict = {word: i for i, word in enumerate(vocabulary)}

    bow_vectors = []
    # 4. Create empty vector in a list, using loop to fill up the vector
    for sentence in sentences:
        sentence_tokens = word_tokenize(sentence.lower())
        current_bow_vector = [0] * len(vocabulary) # Initialize vector with zeros

        for word_in_sentence in sentence_tokens:
            if word_in_sentence in vocabulary_dict:
                index = vocabulary_dict[word_in_sentence]
                current_bow_vector[index] += 1
        bow_vectors.append(current_bow_vector)

    # 5. Return bow vectors and vocabulary in a dictionary
    return bow_vectors, vocabulary_dict

# Example usage with the provided corpus:
bow_vectors_scratch, vocabulary_scratch = create_bow_from_scratch(corpus)

print("\nVocabulary (from scratch):")
print(vocabulary_scratch)

print("\nBoW Vectors (from scratch):")
for i, vec in enumerate(bow_vectors_scratch):
    print(f"Sentence {i+1}: {vec}")
```

# Machine Learning Models

Load the spam ham dataset

```python

# read dataset from a Google Drive File
file_link = 'https://drive.google.com/file/d/1hWX7aB-Cx4OeOsjHTE-Iq1SfxFUpf71t/view?usp=sharing' # the file access must have to be Public
id = file_link.split("/")[-2]
print(id)
new_link = f'https://drive.google.com/uc?id={id}'
print(new_link)
df = pd.read_csv(new_link)
```

```python
df.head()
```

```python
# Split the dataset into 80:20
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['Message'], df['Category'], test_size=0.2, random_state=42)
```

```python
# Preprocessing: convert to small letter
X_train = X_train.str.lower()
X_test = X_test.str.lower()

# Label encoding: ham=0, spam=1
y_train = y_train.map({'ham': 0, 'spam': 1})
y_test = y_test.map({'ham': 0, 'spam': 1})

```

```python
X_train
```

```python
y_train
```

We need to convert text to vector representation. Lets use Bag of word. You can use others like Skipgram, cbow, glove or others too which you will learn in the next class.

```python
bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

### Using Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
# Model Training (Logistic Regression)
lr_model = LogisticRegression()
lr_model.fit(X_train_bow, y_train)
```

```python
from sklearn.metrics import classification_report
# Model Evaluation
y_pred = lr_model.predict(X_test_bow)
print(classification_report(y_test, y_pred))
```

A prediction function for using the model

```python
def predict_spam(sentence):
  # Preprocess the input sentence (convert to lowercase)
  sentence_lower = sentence.lower()
  # Transform the sentence using the fitted TF-IDF vectorizer
  sentence_bow = bow_vectorizer.transform([sentence_lower])
  # Predict using the trained Logistic Regression model
  prediction = lr_model.predict(sentence_bow)
  # Return 'spam' or 'ham' based on the prediction
  if prediction[0] == 1:
    return 'spam'
  else:
    return 'ham'

# Example usage:
test_sentence1 = "URGENT! You have won a 1 week FREE membership in our £100,000 Prize Jackpot!"
prediction1 = predict_spam(test_sentence1)
print(f"'{test_sentence1}' is predicted as: {prediction1}")

test_sentence2 = "Hey, how are you doing today?"
prediction2 = predict_spam(test_sentence2)
print(f"'{test_sentence2}' is predicted as: {prediction2}")
```

```python

test_sentence2 = """Try your own sentences"""
prediction2 = predict_spam(test_sentence2)
print(f"'{test_sentence2}' is predicted as: {prediction2}")
```

### Using Other models

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, f1_score
nb_model = MultinomialNB(alpha = 1) # <-- Laplace/Add one smoothing
nb_model.fit(X_train_bow, y_train)

nb_pred = nb_model.predict(X_test_bow)
nb_accuracy = accuracy_score(y_test, nb_pred)
nb_f1 = f1_score(y_test, nb_pred)
print("NB Accuracy:", nb_accuracy)
print("NB F1 Score:", nb_f1)
print(classification_report(y_test, nb_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier()
rf_model.fit(X_train_bow, y_train)
rf_pred = rf_model.predict(X_test_bow)
rf_accuracy = accuracy_score(y_test, rf_pred)
rf_f1 = f1_score(y_test, rf_pred, average='macro')
print("RF Accuracy:", rf_accuracy)
print("RF F1 Score:", rf_f1)
print(classification_report(y_test, rf_pred))
```
