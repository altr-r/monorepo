# Task 2

```python
df = df.drop(columns=["UserName", "ScreenName", "Location", "TweetAt"])
```

```python
df
```

```python
sentiment_map = {
  "Extremely Positive": 0,
  "Extremely Negative": 1,
  "Positive": 2,
  "Neutral": 3,
  "Negative": 4
}

df["Sentiment"] = df["Sentiment"].map(sentiment_map)
```

```python
df
```

```python
df["OriginalTweet"] = df["OriginalTweet"].str.lower()
```

```python
df
```

```python
from nltk.corpus import stopwords
stop_words = stopwords.words("english")
```

```python
original_tweet = df["OriginalTweet"]
filtered_sentences = []
for i in original_tweet:
  filtered_word = []
  for j in i.split():
    if j not in stop_words:
      filtered_word.append(j)
  filtered_sentences.append(" ".join(filtered_word))
df['OriginalTweet'] = filtered_sentences
df
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

original_tweet = df["OriginalTweet"]
stemmed_sentences = []
for i in original_tweet:
  stemmed_word = []
  for j in i.split():
    stemmed_word.append(stemmer.stem(j))
  stemmed_sentences.append(" ".join(stemmed_word))
df['OriginalTweet'] = stemmed_sentences

df
```

# Task 3

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
x = df["OriginalTweet"]
y = df["Sentiment"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
  x,
  y,
  test_size=0.2,
  random_state=42,
  stratify=y
)
```

```python
vec = TfidfVectorizer(max_features=12000, ngram_range=(1, 3))
```

```python
x_train_vec = vec.fit_transform(x_train)
x_test_vec = vec.transform(x_test)

X_train = x_train_vec.toarray().astype("float32")
X_test = x_test_vec.toarray().astype("float32")

X_train.shape
```

```python
from tensorflow.keras.utils import to_categorical
y_train_onehot = to_categorical(y_train, num_classes = 5)
y_test_onehot = to_categorical(y_test, num_classes = 5)
```

# Task 4

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

import matplotlib.pyplot as plt
```

```python
modelA = Sequential([
    Input(shape= (X_train.shape[1],)),
    Dense(256, activation="leaky_relu"),
    Dropout(0.2),
    Dense(128, activation="relu"),
    Dropout(0.2),
    Dense(64, activation="relu"),
    Dense(5, activation="softmax")
])
```

```python
modelA.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
historyA = modelA.fit(
    X_train,
    y_train_onehot,
    validation_split=0.15,
    epochs=50,
    batch_size=32,
    verbose=1
)
```

```python
def plot_history(history, title):

    plt.figure(figsize=(12,4))

    # Accuracy
    plt.subplot(1,2,1)
    plt.plot(history.history["accuracy"])
    plt.plot(history.history["val_accuracy"])
    plt.title(title + " Accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.legend(["Train","Validation"])


    # Loss
    plt.subplot(1,2,2)
    plt.plot(history.history["loss"])
    plt.plot(history.history["val_loss"])
    plt.title(title + " Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend(["Train","Validation"])

    plt.show()
```

```python
plot_history(historyA, "Model A")
```

```python
pred_A = modelA.predict(
    X_test.toarray()
)

pred_A_classes = pred_A.argmax(axis=1)

true_classes = y_test_onehot.argmax(axis=1)
```

```python
print(
    classification_report(
        true_classes,
        pred_A_classes
    )
)
```
